```python
def chatbot(user_message):

    user_message = user_message.lower()

    if user_message == "hello":
        return "Hi! How can I help you?"

    elif user_message == "hi":
        return "Hello! Nice to meet you."

    elif user_message == "how are you":
        return "I'm fine! Thank you."

    elif user_message == "what is your name":
        return "I am Python Chatbot."

    elif user_message == "what can you do":
        return "I can answer simple questions."

    elif user_message == "thank you":
        return "You're welcome!"

    elif user_message == "bye":
        return "Goodbye! Have a nice day."

    else:
        return "Sorry! I don't understand."


print("================================")
print("         PYTHON CHATBOT")
print("================================")

print("Chatbot: Hello! Welcome to my chatbot.")
print("Chatbot: Type 'bye' to exit.")

while True:

    user_input = input("\nYou: ")

    response = chatbot(user_input)

    print("Chatbot:", response)

    if user_input.lower() == "bye":
        break

print("\nChatbot session ended!")
```
