```python
stock_prices = {
    "AAPL": 180,
    "TSLA": 250,
    "GOOGL": 140,
    "MSFT": 420,
    "AMZN": 190
}

portfolio = {}
total_investment = 0

print("================================")
print("     STOCK PORTFOLIO TRACKER")
print("================================")

print("\nAvailable Stocks:")

for stock, price in stock_prices.items():
    print(stock, ": $", price)

num_stocks = int(input("\nHow many stocks do you want to buy? "))

for i in range(num_stocks):

    stock_name = input("\nEnter stock name: ").upper()

    if stock_name in stock_prices:

        quantity = int(input("Enter quantity: "))

        if quantity <= 0:
            print("Quantity must be positive!")
            continue

        investment = stock_prices[stock_name] * quantity

        portfolio[stock_name] = {
            "Quantity": quantity,
            "Price": stock_prices[stock_name],
            "Investment": investment
        }

        total_investment += investment

        print("Stock added successfully!")

    else:
        print("Stock not available!")

print("\n========== YOUR PORTFOLIO ==========")

for stock, details in portfolio.items():

    print("\nStock:", stock)
    print("Quantity:", details["Quantity"])
    print("Price: $", details["Price"])
    print("Investment: $", details["Investment"])

print("\n================================")
print("Total Investment: $", total_investment)
print("================================")

with open("portfolio.txt", "w") as file:

    file.write("STOCK PORTFOLIO REPORT\n")
    file.write("======================\n")

    for stock, details in portfolio.items():

        file.write("\nStock: " + stock)
        file.write("\nQuantity: " + str(details["Quantity"]))
        file.write("\nPrice: $" + str(details["Price"]))
        file.write("\nInvestment: $" + str(details["Investment"]))
        file.write("\n")

    file.write("\nTotal Investment: $" + str(total_investment))

print("\nPortfolio saved successfully in portfolio.txt")
```
